6 & 8. WHICH COLUMNS AND ROWS HAVE MISSING (NULL) VALUES?

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())
```

7. WHICH ARE THE FEATURES AND WHICH IS THE TARGET?

```python
target = "Species"
features = df.drop(columns=["Species", "Id"])

print("Target:", target)
print("Example feature columns:", features.columns.tolist())
print("Target values:", df[target].unique())
```

**Answer** :
Target: Species. This is what we want to predict, and it has 3 values: Iris-setosa, Iris-versicolor and Iris-virginica.

Features: these are the input columns used to predict the target:

1. SepalLengthCm
2. SepalWidthCm
3. PetalLengthCm
4. PetalWidthCm

Id is not a feature. It is only a row number and says nothing about the flower, so it was removed along with the target.

9. WHAT USEFUL QUESTION CAN WE ASK? DOES THIS DATASET FIT?

**Useful Question** : What is the average petal length for each Iris species, and which species has the longest average petal length?

**Does the Dataset Fit?** : Yes, the Iris dataset fits this question because it contains the species names and petal length measurements needed to calculate and compare the average petal length of each species.

```python
avg_petal = df.groupby("Species")["PetalLengthCm"].mean()
print("Average petal length (cm) for each species:")
print(avg_petal.round(2))
```
